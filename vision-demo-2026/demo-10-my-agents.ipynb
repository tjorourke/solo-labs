# Part 10 — My agents (code-free)

A non-developer creates an agent from a prompt, optional Telco/IT skills, and the MCP tools they allow. No code. YAML is generated and deployed to kagent on mesh1. Skills live in the catalogue; tools are an allow-list (nothing runs until it is ticked).

Open the console (same `demo-console/run.sh` as the other cards):

**http://localhost:8900/agents**

kagent is not on mesh1 until you run `demo-scripts/agentregistry/setup-mesh1.sh`. The console still generates the manifests. Deploy again after that standup.
