## 1. One endpoint, many providers

Clients call one OpenAI-compatible endpoint and ask for a **corporate model name**. The gateway extracts the model from the request body *before* routing, matches it to the right provider, and each Backend pins the real deployment. Clients never hold provider credentials and never learn provider endpoints.

- `corporate-fast` → Azure OpenAI (`gpt-5-nano`)
- `corporate-secure` → AWS Bedrock (`claude-haiku-4-5`)
- `premium-reasoning` → Anthropic (`claude-sonnet-4-5`, via a native model alias)
- `corporate-premium` → Anthropic (`claude-opus-5`, the top reasoning tier)
- anything else → the corporate default, Azure OpenAI

````bash
kubectl apply -f - <<'YAML'
# Azure OpenAI: the deployment is pinned at the gateway
apiVersion: enterpriseagentgateway.solo.io/v1alpha1
kind: EnterpriseAgentgatewayBackend
metadata:
  name: azure-gpt5
  namespace: agentgateway-system
spec:
  ai:
    provider:
      host: azure-openai.ai-models.svc.cluster.local
      port: 8000
      path: /v1/chat/completions
      openai:
        model: gpt-5-nano
  policies:
    auth:
      passthrough: {}
---
# AWS Bedrock, same shape
apiVersion: enterpriseagentgateway.solo.io/v1alpha1
kind: EnterpriseAgentgatewayBackend
metadata:
  name: bedrock-haiku
  namespace: agentgateway-system
spec:
  ai:
    provider:
      host: bedrock.ai-models.svc.cluster.local
      port: 8000
      path: /v1/chat/completions
      openai:
        model: claude-haiku-4-5
  policies:
    auth:
      passthrough: {}
---
# Anthropic: modelAliases maps the corporate name to the real model, and the
# gateway translates OpenAI-shape requests to /v1/messages.
apiVersion: enterpriseagentgateway.solo.io/v1alpha1
kind: EnterpriseAgentgatewayBackend
metadata:
  name: anthropic-claude
  namespace: agentgateway-system
spec:
  ai:
    provider:
      anthropic: {}
  policies:
    auth:
      secretRef:
        name: anthropic-secret
    ai:
      modelAliases:
        premium-reasoning: claude-sonnet-4-5
        corporate-premium: claude-opus-5
YAML
````

Two mechanisms make the corporate names work. `modelAliases` (on the Anthropic Backend above) maps a friendly name to the real model within a backend. For picking **which provider** a name goes to, a `PreRouting` transformation extracts the `model` field from the JSON body into a header, and a standard `HTTPRoute` matches on it:

````bash
kubectl apply -f - <<'YAML'
# Runs before route selection, so routes can match on the body-derived header
apiVersion: agentgateway.dev/v1alpha1
kind: AgentgatewayPolicy
metadata:
  name: extract-model
  namespace: agentgateway-system
spec:
  targetRefs:
  - name: ai-gateway
    group: gateway.networking.k8s.io
    kind: Gateway
  traffic:
    phase: PreRouting
    transformation:
      request:
        set:
        - name: x-model-name
          value: "default(json(request.body).model, '')"
---
apiVersion: gateway.networking.k8s.io/v1
kind: HTTPRoute
metadata:
  name: models
  namespace: agentgateway-system
spec:
  parentRefs:
  - name: ai-gateway
  rules:
  - matches:
    - path: { type: PathPrefix, value: /models }
      headers:
      - { type: Exact, name: x-model-name, value: corporate-fast }
    backendRefs:
    - { name: azure-gpt5, group: enterpriseagentgateway.solo.io, kind: EnterpriseAgentgatewayBackend }
    timeouts: { request: 120s }
  - matches:
    - path: { type: PathPrefix, value: /models }
      headers:
      - { type: Exact, name: x-model-name, value: corporate-secure }
    backendRefs:
    - { name: bedrock-haiku, group: enterpriseagentgateway.solo.io, kind: EnterpriseAgentgatewayBackend }
    timeouts: { request: 120s }
  - matches:
    - path: { type: PathPrefix, value: /models }
      headers:
      - { type: Exact, name: x-model-name, value: premium-reasoning }
    backendRefs:
    - { name: anthropic-claude, group: enterpriseagentgateway.solo.io, kind: EnterpriseAgentgatewayBackend }
    timeouts: { request: 120s }
  - matches:
    - path: { type: PathPrefix, value: /models }
      headers:
      - { type: Exact, name: x-model-name, value: corporate-premium }
    backendRefs:
    - { name: anthropic-claude, group: enterpriseagentgateway.solo.io, kind: EnterpriseAgentgatewayBackend }
    timeouts: { request: 120s }
  # anything else falls through to the corporate default provider
  - matches:
    - path: { type: PathPrefix, value: /models }
    backendRefs:
    - { name: azure-gpt5, group: enterpriseagentgateway.solo.io, kind: EnterpriseAgentgatewayBackend }
    timeouts: { request: 120s }
YAML
````

Same request shape, four different providers. Watch the `model` in each response:

````bash
for M in corporate-fast corporate-secure premium-reasoning corporate-premium anything-else; do
  echo "== requested: $M =="
  curl -sS -m 60 "http://$GATEWAY/models/v1/chat/completions" \
    -H 'content-type: application/json' \
    -d '{"model":"'$M'","messages":[{"role":"user","content":"In exactly 5 words: what is a service mesh?"}],"max_tokens":400}' \
    | jq -c '{served_by: .model, answer: (.choices[0].message.content | if length > 70 then .[:70]+"…" else . end)}'
done
````

## 2. Failover and load balancing between providers

A single Backend can declare **priority groups**, ordered tiers of providers. Providers inside a group are load-balanced and evicted independently; a `health` policy evicts on 5xx or 429, and traffic falls to the next group only when the current group is exhausted. No client retries, no SDK changes.

Here the primary is the corporate default, Azure OpenAI, and the fallback group holds AWS Bedrock and Anthropic together. When the primary dies, the gateway balances requests across both from the same endpoint.

## 6. Virtual keys for services and CI

Not every caller has a user JWT. CI jobs, batch pipelines and agents authenticate with **virtual keys**: opaque API keys stored as labelled Kubernetes Secrets, each carrying identity metadata. The gateway authenticates the key and the metadata flows into budgets and cost attribution exactly like JWT claims. Platform teams mint keys without ever touching provider credentials.

Spend limits are **declarative**: an `EnterpriseAgentgatewayBudget` states the intent ("this key gets 100 tokens a day, block on overage") and the controller compiles the enforcement, no descriptors to hand-write. The day's spend is tracked durably per key, a restart never resets an allowance. Budgets also come in `USD` (priced by the §7 cost catalog), `Week`/`Month`/`Year` windows, and `Audit` mode to observe before enforcing.

The proxy exports `agentgateway_gen_ai_client_token_usage_sum` and `agentgateway_gen_ai_client_cost_usd_total`, and thanks to the §3 metrics policy every series carries `user`, `team` and `business_unit` labels.
````bash
: "${NS:=agentgateway-system}"
pkill -f "port-forward deploy/ai-gateway 15020" 2>/dev/null
kubectl -n $NS port-forward deploy/ai-gateway 15020:15020 >/dev/null 2>&1 &
sleep 3
sum_by() {  # <metric> <label[,label]> — sum a counter by one or more labels
  curl -s localhost:15020/metrics | awk -v metric="$1" -v labels="$2" '
    $0 ~ "^"metric"{" {
      n = split(labels, want, ","); split($0, kv, /[{}]/); key = ""
      for (i = 1; i <= n; i++) {
        if (match(kv[2], want[i]"=\"[^\"]*\"")) {
          v = substr(kv[2], RSTART, RLENGTH); sub(/^[^"]*"/, "", v); sub(/"$/, "", v)
        } else v = "unknown"
        key = key (i>1 ? " · " : "") v
      }
      sums[key] += $NF
    }
    END { for (k in sums) printf "  %-55s %.6f\n", k, sums[k] }'
}
echo "== tokens by business unit ==";  sum_by agentgateway_gen_ai_client_token_usage_sum business_unit
echo; echo "== tokens by team ==";     sum_by agentgateway_gen_ai_client_token_usage_sum team
echo; echo "== realised USD by user and model =="; sum_by agentgateway_gen_ai_client_cost_usd_total "user,gen_ai_request_model"
echo; echo "== every model priced against the catalog (Exact = priced) =="
curl -s localhost:15020/metrics | grep '^agentgateway_cost_catalog_lookups_total' \
  | grep -oE 'status="[^"]*"|gen_ai_request_model="[^"]*"' | paste - - | sort | uniq -c | sed 's/^/  /'
pkill -f "port-forward deploy/ai-gateway 15020" 2>/dev/null
````

### Govern it: per-tool authorization from identity claims

Now lock the same `/mcp` endpoint down in place, the URL agents already point at doesn't change. The same JWTs as the LLM sections: a CEL rule decides which tools each identity can *see* (`tools/list`) and *call* (`tools/call`). `printenv` dumps the server's environment, so it's for engineering only:

````bash
JWKS=$(jq -c . demo-scripts/.jwt/jwks.json)
kubectl apply -f - <<YAML
apiVersion: enterpriseagentgateway.solo.io/v1alpha1
kind: EnterpriseAgentgatewayPolicy
metadata:
  name: mcp-jwt
  namespace: agentgateway-system
spec:
  targetRefs:
  - group: gateway.networking.k8s.io
    kind: HTTPRoute
    name: mcp
  traffic:
    jwtAuthentication:
      mode: Strict
      providers:
      - issuer: https://idp.demo.example/
        jwks:
          inline: '$JWKS'
---
apiVersion: agentgateway.dev/v1alpha1
kind: AgentgatewayPolicy
metadata:
  name: mcp-tool-authz
  namespace: agentgateway-system
spec:
  targetRefs:
  - group: gateway.networking.k8s.io
    kind: HTTPRoute
    name: mcp
  backend:
    mcp:
      authorization:
        action: Allow
        policy:
          matchExpressions:
          - 'mcp.tool.name != "printenv" || jwt.groups.exists(g, g == "engineering")'
YAML
sleep 5
echo "-- tools visible to bob (finance): --";  mcp_tools "$BOB"
echo; echo "-- tools visible to alice (engineering): --"; mcp_tools "$ALICE"
echo; echo -n "-- anonymous initialize: "; status -X POST "http://$GATEWAY/mcp" -H 'content-type: application/json' -d '{"jsonrpc":"2.0","id":1,"method":"initialize","params":{"protocolVersion":"2024-11-05","capabilities":{},"clientInfo":{"name":"x","version":"1"}}}'
````

## Where this goes next

Everything above was one gateway, one set of CRDs, and identity threaded from the JWT (or virtual key) through routing, access, quota, budget and cost. What didn't fit in this part:

- **Agent identity and on-behalf-of token exchange**: the gateway ships an RFC 8693 STS and native Entra OBO, so a downstream API sees "agent X acting for user Y" instead of a shared service account. That needs a real IdP tenant and install-time validator config; the `agentgateway-token-exchange-kind` and kagent enterprise OBO labs cover it end to end.
- **Guardrails**: prompt guard, WAF for agentic traffic shapes, and external guardrail engines over ExtProc (`agentic-external-guardrail-kind`).
- **Semantic caching, streaming, embeddings and the inference extension** (Part 6 covers KV-cache-aware routing to self-hosted pools).
